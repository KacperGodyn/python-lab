```python
import multiprocessing
import time
from lab2_functions import calculate_power_sum

if __name__ == "__main__":
    numbers = range(1, 10001)
    
    start_time = time.time()
    
    with multiprocessing.Pool() as pool:
        results = pool.map(calculate_power_sum, numbers)
        
    end_time = time.time()
    
    print(f"Przetworzono {len(results)} elementów.")
    print(f"Czas wykonania (multiprocessing): {end_time - start_time:.4f} s")
```
