```python
import requests
import time
import concurrent.futures

CAT_API_URL = "https://catfact.ninja/fact"

def fetch_fact():
    """Funkcja pomocnicza pobierająca jeden fakt o kotach."""
    try:
        response = requests.get(CAT_API_URL, timeout=5)
        return response.json().get('fact')
    except Exception as e:
        return f"Błąd pobierania: {e}"

print("--- Pobieranie sekwencyjne ---")
start_time_seq = time.time()

facts_seq = []
for _ in range(20):
    facts_seq.append(fetch_fact())

end_time_seq = time.time()
seq_duration = end_time_seq - start_time_seq
print(f"Czas wykonania sekwencyjnego: {seq_duration:.4f} sekund")

print("\n--- Pobieranie wielowątkowe (ThreadPoolExecutor) ---")
start_time_thread = time.time()

facts_thread = []
with concurrent.futures.ThreadPoolExecutor(max_workers=10) as executor:
    results = executor.map(lambda _: fetch_fact(), range(20))
    facts_thread = list(results)

end_time_thread = time.time()
thread_duration = end_time_thread - start_time_thread
print(f"Czas wykonania wielowątkowego: {thread_duration:.4f} sekund")

print("\n--- Podsumowanie ---")
print(f"Metoda sekwencyjna trwała: {seq_duration:.4f} s")
print(f"Metoda wielowątkowa trwała: {thread_duration:.4f} s")
print(f"Zysk czasowy (wielowątkowość szybsza o): {seq_duration - thread_duration:.4f} s")
```
