```python
import queue
import threading
import time

def producer(q, limit):
    for i in range(1, limit + 1):
        q.put(i)

def consumer_1(q, stop_event):
    while not (stop_event.is_set() and q.empty()):
        try:
            item = q.get(timeout=0.05)
            if item % 2 == 0:
                print(f"Konsument 1 (parzyste): {item}")
                q.task_done()
            else:
                q.put(item)
                q.task_done()
                time.sleep(0.001)
        except queue.Empty:
            continue

def consumer_2(q, stop_event):
    while not (stop_event.is_set() and q.empty()):
        try:
            item = q.get(timeout=0.05)
            if item % 2 != 0:
                print(f"Konsument 2 (nieparzyste): {item}")
                q.task_done()
            else:
                q.put(item)
                q.task_done()
                time.sleep(0.001)
        except queue.Empty:
            continue

q = queue.Queue()
stop_event = threading.Event()

prod = threading.Thread(target=producer, args=(q, 20))
cons1 = threading.Thread(target=consumer_1, args=(q, stop_event))
cons2 = threading.Thread(target=consumer_2, args=(q, stop_event))

prod.start()
cons1.start()
cons2.start()

prod.join()
q.join()
stop_event.set()

cons1.join()
cons2.join()
```
