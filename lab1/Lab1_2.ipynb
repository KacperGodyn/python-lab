```python
import functools
from datetime import datetime
import time

def logger(filename):
    def decorator(func):
        @functools.wraps(func)
        def wrapper(*args, **kwargs):
            start_time = time.perf_counter()
            execution_date = datetime.now()
            try:
                return func(*args, **kwargs)
            finally:
                duration = time.perf_counter() - start_time
                with open(filename, 'a', encoding='utf-8') as f:
                    f.write(f"Funkcja: {func.__name__}, Data: {execution_date.strftime('%Y-%m-%d %H:%M:%S')}, Czas wykonania: {duration:.6f}s\n")
        return wrapper
    return decorator

# Test:
@logger("funkcja.log")
def testowa_funkcja(sekundy):
    time.sleep(sekundy)
    return "zrobione"

testowa_funkcja(0.1)
```
