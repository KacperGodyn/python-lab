```python
nameSurname = input("Podaj swoje imię i nazwisko: ")
```

```python
stringArr = nameSurname.split(" ")

initials = []

for string in stringArr:
    string = string.capitalize()
    initials.append((string[0]))

print(".".join(initials) + ".")
```

```python
name = stringArr[0].capitalize()

i = len(name) - 1
nameBackwards = []

while i >= 0:
    nameBackwards.append(name[i])
    i -= 1

print("".join(nameBackwards))
```

```python
nameSurnameLength = 0

for string in stringArr:
    string = string.strip()
    nameSurnameLength += len(string)

print(nameSurnameLength)
```
