```python
zakupy = [
    {"nazwa": "mleko", "cena": 4.50, "ilość": 2},
    {"nazwa": "chleb", "cena": 5.00, "ilość": 1},
    {"nazwa": "masło", "cena": 7.99, "ilość": 1},
    {"nazwa": "jajka", "cena": 12.50, "ilość": 1},
    {"nazwa": "mleko", "cena": 4.50, "ilość": 3},
]
```

```python
całkowitaCenaZakupów = []

for produkt in zakupy:
    całkowitaCenaPerProdukt = produkt["cena"] * produkt["ilość"]
    całkowitaCenaZakupów.append(całkowitaCenaPerProdukt)
    print(f"całkowita cena za {produkt["nazwa"]} wynosi {całkowitaCenaPerProdukt}")


print(f"""
całkowita cena zakupów: {sum(całkowitaCenaZakupów)}""")
```

```python
ceny = []

for produkt in zakupy:
    ceny.append(produkt["cena"])

for produkt in zakupy:
    if(max(ceny) == produkt["cena"]):
        maxProdukt = produkt["nazwa"]

print(f"{maxProdukt} to najdroższy produkt kosztujący {max(ceny)} za szt.")
```

```python
produkty = set()

for produkt in zakupy:
    produkty.add(produkt["nazwa"])

print(list(produkty))
```

```python
może_i_źle_ale_za_to_drogo = [produkt["nazwa"] for produkt in zakupy if produkt["cena"] > 5]

print(f"""produkty droższe niż 5 zł: 
{", ".join(może_i_źle_ale_za_to_drogo)}""")
```
