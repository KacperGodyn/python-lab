```python
import random

while True:
    print("Myślę o liczbie...")
    target = random.randint(1,100)

    while True:
        user_num = input("Wybierz liczbę: ")
        user_num = int(user_num)

        if user_num == target:
            print(f"Udało się! Podana liczba {user_num} to {target}, o której myślałem")
            break
        elif user_num > target:
            print(f"Podana liczba ({user_num}) jest za duża!")
        elif user_num < target:
            print(f"Podana liczba ({user_num}) jest za mała!")
    
    replay = input("Czy chcesz zagrać ponownie? y/n")

    if replay.lower() == "n":
        break
```
