```python
import csv
```

```python
contact_headers = ["name", "surname", "phone_number", "email"]
```

```python
def contact_add(contact, contacts):
    """Dodaje kontakt in-memory"""

    contact_details = contact.split(",")
    contact_details[2] = int(contact_details[2])

    if len(contact_details) == len(contact_headers):
        if contact_details[2] != 0 and len(str(contact_details[2])) == 9:

            if "@" in contact_details[3]:

                i = 0
                temp_map = {}
                for header in contact_headers:
                    temp_map[header] = str(contact_details[i]).strip()
                    i += 1
                contacts.append(temp_map)
                print(f"Dodano kontakt: {temp_map}")
            else:
                print("Zły format, email nie zawiera znaku '@'!")
        else:
            print("Zły format, numer jest równy zeru lub jego długość nie jest równa 9")
    else:
        print("Za mała/za duża ilość parametrów dla kontaktu!")
```

```python
def contact_find(target, contacts):
    """Znajduje kontakt po imieniu/nazwisku (target), jeśli target = \"*\", to wyświetla wszystkie kontakty"""
    contact_found = False
    i = 0
    for contact in contacts:
        if target == "*":
            contact_found = True
            print(contact)
        elif target.upper() in contact["name"].upper() or target.upper() in contact["surname"].upper():
            contact_found = True
            print(contact)
        i += 1

    if not contact_found:
        print("Nie znaleziono kontaktu!")
```

```python
def contact_del(id, contacts):
    """Usuwa kontakt"""
    deleted_contact = contacts.pop(int(id))
    print(f"Pomyślnie usunięto kontakt: {deleted_contact}")
```

```python
def contact_write_to_file(file, contacts):
    """Zapisuje kontakty do pliku z konktaktów in-memory"""
    with open(file, "w", newline="") as wf:
        csv_writer = csv.DictWriter(wf, fieldnames=contact_headers)
        csv_writer.writeheader()

        for contact in contacts:
            csv_writer.writerow(contact)
```

```python
def contact_read_from_file(file, contacts):
    """Wczytuje kontakty z pliku do konktaktów in-memory"""
    with open(file, "r") as rf:
        csv_reader = csv.reader(rf)

        next(csv_reader)

        for line in csv_reader:            
            contact_add(",".join(line), contacts)
```

```python
contacts = []

while True:

    user_input = input("""Wybierz opcję:
1. Dodaj kontakt (imię, nazwisko, telefon, email)
  2. Wyświetl wszystkie kontakty
  3. Wyszukaj kontakt (po imieniu lub nazwisku)
  4. Usuń kontakt
  5. Zapisz kontakty do pliku (CSV)
  6. Wczytaj kontakty z pliku (CSV)
  7. Wyjdź
    """)

    user_input = int(user_input)

    if (user_input == 1):
        contact = input("Dodaj nowy kontakt(format imię, nazwisko, telefon, email): ")
        contact_add(contact, contacts)
    elif (user_input == 2):
        contact_find("*", contacts)
    elif (user_input == 3):
        target_contact = input("Podaj imię lub nazwisko kontaktu, który chcesz znaleźć: ")
        contact_find(target_contact, contacts)
    elif (user_input == 4):
        contact_id_to_delete = input("Podaj index kontaktu do usunięcia")
        contact_del(contact_id_to_delete, contacts)
    elif (user_input == 5):
        file = input("Podaj ścieżkę do pliku docelowego: ")
        contact_write_to_file(file, contacts)
    elif (user_input == 6):
        file = input("Podaj ścieżkę do pliku docelowego: ")
        contact_read_from_file(file, contacts)
    elif (user_input == 7):
        break
```
