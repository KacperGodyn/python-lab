```python
import csv
import statistics

with open("studenci.csv", "r") as studenci_file:
    studenci_reader = csv.DictReader(studenci_file)

    oceny = []

    with open("raport.txt", "w") as raport_file:
        raport_file.write("0 - srednia ocen\n\n")
        raport_file.write("uczniowie z najwyzsza ocena:\n")

        for student in studenci_reader:
            oceny.append(int(student["grade"]))

            if (int(student["grade"]) == max(oceny)):
                raport_file.write(f"{student["name"]}: ({student["grade"]})\n")
        
        raport_file.seek(0)
        raport_file.write(str(statistics.mean(oceny)))
```
