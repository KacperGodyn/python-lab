```python
def tekst_na_slowa(tekst):
    """Zamienia tekst na listę słow"""
    slowa = tekst.split(" ")
    return slowa
```

```python
def policz_slowa(tekst):
    """Zwraca liczbę słów w tekście."""
    return len(tekst_na_slowa(tekst))
```

```python
def najdluzsze_slowo(tekst):
    """Zwraca najdłuższe słowo w tekście."""
    dlugosci_slow = []

    for slowo in tekst_na_slowa(tekst):
        dlugosci_slow.append(len(slowo))

    return max(dlugosci_slow)

najdluzsze_slowo("wwafa sd zcv e wawawafeafs")
```

```python
def czestotliwosc_slow(tekst):
    """Zwraca słownik {słowo: liczba_wystąpień}."""

    czestotliwosc = {}

    for slowo in tekst_na_slowa(tekst):
        czestotliwosc[slowo] = tekst_na_slowa(tekst).count(slowo)

    return czestotliwosc
```

```python
def cenzura(tekst, zakazane_slowa):
    
    tekst_na_slowa_var = tekst_na_slowa(tekst)

    i = 0
    for slowo in tekst_na_slowa(tekst):
        for zakazane_slowo in zakazane_slowa:
            if zakazane_slowo.upper() in slowo.upper():
                tekst_na_slowa_var[i] = "***"
        i += 1
    
    result = " ".join(tekst_na_slowa_var)

    return result
```

```python
tekst = "Bazylika katedralna pw. Chrystusa Króla w Reykjavíku, Kościół Landakot (isl. Basilika Krists konungs, Landakotskirkja) – rzymskokatolicka katedra znajdująca się w zachodniej części Reykjavíku w dzielnicy Landakot, siedziba biskupów, obejmującej całe terytorium Islandii, diecezji reykjavíckiej, wznoszona od końca I wojny światowej do 1929 według projektu Guðjóna Samúelssona[1] w stylu neogotyckim[2], określana również jak Kristskirkja (Kościół Chrystusa). Katedra poświęcona jest Chrystusowi Królowi, jej patronami są również Najświętsza Maria Panna Bogurodzica, św. Józef, św. Torlak i św. Jón."
```

```python
policz_slowa(tekst)
```

```python
najdluzsze_slowo(tekst)
```

```python
czestotliwosc_slow(tekst)
```

```python
zakazane_slowa = ["św.", "rzymskokatolicka", "diecezji", "katedra", "kościół"]

print(tekst)
cenzura(tekst, zakazane_slowa)
```
